```python
import math
def kiem_tra_snt(n):
    if n<2:
        return False
    for k in range(2,int(math.isqrt(n)+1)):
        if n%k==0:
            return False    
    return True
n=int(input())
if kiem_tra_snt(n):
    print("n la snt")
else: 
    print('deophai')

xoadc=False
s=str(n)
ans=[]
for i in range(len(s)):
    k=s[:i]+s[i+1:]
    if k and kiem_tra_snt(int(k)):
        xoadc=True
        ans.append(int(k))
if xoadc==True:
    print('xoa dc: ',ans)
else:
    print('ko xoa dc')

    
```
